# ***Aerofit Treadmill Customer Analysis***

## ***Problem Statement***
*Aerofit, a leading fitness equipment brand, aims to optimize its marketing and sales strategies by identifying distinct customer profiles for its treadmill products: KP281 (1500 dollars, entry-level), KP481 (1750 dollars, mid-level), and KP781 (2500 dollars, advanced). This analysis seeks to explore customer characteristics—such as age, gender, income, education, fitness level, usage frequency, and miles run—to determine how they influence treadmill purchases. Key questions include: Who are the typical buyers of each treadmill model? How do demographic and behavioral factors affect product choice? What are the conditional and marginal probabilities of purchasing each treadmill given specific customer traits? By employing descriptive analytics and probability analysis, this study will deliver actionable insights to recommend the right treadmill to new customers and enhance targeted marketing efforts.*

# ***Converting Categorical Attributes***

```python
# Converting categorical columns to 'category' type for efficiency
df['Product'] = df['Product'].astype('category')
df['Gender'] = df['Gender'].astype('category')
df['MaritalStatus'] = df['MaritalStatus'].astype('category')

print("Data Types After Conversion:")
print("----------------------------")
print(df.dtypes)
```

# ***Binning features into categories***

```python
# Binning Age into categories
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 25, 35, 50, float('inf')],
                         labels=['Young (18-25)', 'Adult (26-35)', 'Middle-Aged (36-50)', 'Senior (>50)'],
                         include_lowest=True).astype('category')
```

```python
# Binning Education into categories
df['Education_Group'] = pd.cut(df['Education'], bins=[0, 14, 16, float('inf')],
                              labels=['High School (≤14)', 'College (15-16)', 'Advanced (>16)'],
                              include_lowest=True).astype('category')
```

```python
# Binning Income into categories
df['Income_Group'] = pd.cut(df['Income'], bins=[0, 40000, 60000, 80000, float('inf')],
                            labels=['Low (<40K)', 'Medium (40K-60K)', 'High (60K-80K)', 'Very High (>80K)'],
                            include_lowest=True).astype('category')
```

```python
# Binning Usage into categories
df['Usage_Group'] = pd.cut(df['Usage'], bins=[0, 2, 4, 7],
                           labels=['Low (1-2)', 'Moderate (3-4)', 'High (5-7)'],
                           include_lowest=True).astype('category')
```

```python
# Binning Fitness into categories
df['Fitness_Group'] = pd.cut(df['Fitness'], bins=[0, 2, 3, 5],
                            labels=['Low (1-2)', 'Moderate (3)', 'High (4-5)'],
                            include_lowest=True).astype('category')
```

```python
# Binning Miles into categories
df['Miles_Group'] = pd.cut(df['Miles'], bins=[0, 50, 100, 200, float('inf')],
                           labels=['Light (0-50)', 'Moderate (51-100)', 'Active (101-200)', 'Enthusiast (>200)'],
                           include_lowest=True).astype('category')
```

```python
print("Data Types After Conversion:")
print("----------------------------")
print(df.dtypes)
```

***Observations:***

*   *Created categorical bins for Age, Education, Income, Usage, Fitness, and Miles to enable detailed EDA.*
*   *All new columns are categorical, suitable for countplots and contingency tables.*

# ***Contingency Table - Product vs Gender***

```python
# Creating and normalizing crosstab for Product vs Gender
gender_crosstab = pd.crosstab(df['Product'], df['Gender'], normalize='all').round(2)
print("Product vs Gender (Marginal Probabilities):")
print("-------------------------------------------")
print(gender_crosstab)

# Conditional Probability: P(Product | Gender)
gender_conditional = pd.crosstab(df['Product'], df['Gender'], normalize='columns').round(2)
print("\nConditional Probability (Product | Gender):")
print("-------------------------------------------")
print(gender_conditional)
```

***Observations:***

*   *Marginal: 24% of purchases are KP281 by males; 8% are KP781 by females.*
*   *Conditional: P(KP781 | Male) = 0.32, P(KP781 | Female) = 0.09, showing males are more likely to buy KP781.*


# ***Contingency Table - Product vs Age Group***

```python
age_crosstab = pd.crosstab(df['Product'], df['Age_Group'], normalize='all').round(2)
print("Product vs Age Group (Marginal Probabilities):")
print("------------------------------------------------------------")
print(age_crosstab)

# Conditional Probability: P(Product | Age Group)
age_conditional = pd.crosstab(df['Product'], df['Age_Group'], normalize='columns').round(2)
print("\nConditional Probability (Product | Age Group):")
print("------------------------------------------------------------")
print(age_conditional)
```

***Observations:***
